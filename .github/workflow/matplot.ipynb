```python
#line plot
%pip install matplotlib
import matplotlib.pyplot as plt
x = [1, 2, 3, 4, 5,]
y = [2, 4, 6, 8, 10 ]

plt.plot(x, y)
plt.title("Simple Line Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()
```

```python
#  scatter plot
x = [5, 7, 8, 10, 12]
y = [10, 14, 15, 18, 20]

plt.scatter(x, y, color="red", marker="o")
plt.title("scatter plot Example")
plt.xlabel("x Values")
plt.ylabel("y values")
plt.show()
```

```python
# vertical bar chart
x = ['apple', 'banana', 'cherry', 'mango']
y = [30, 45, 10, 23]

plt.bar(x, y, color='green')
plt.title("tn fruit sales")
plt.xlabel("fruit")
plt.ylabel("quntati solt")
plt.show()
```

```python
#plt.brch
x = ['python', 'java', 'c++', 'javascript', 'html']
y = [98, 76, 54, 67, 54]

plt.barh(x, y, color='blue')
plt.title("programing language useage")
plt.xlabel("score")
plt.ylabel("Language")
plt.show()
```

```python
#histogram
import numpy as np

data = np.random.randn(1000)

plt.hist(data, bins=20, color='green', edgecolor='black')
plt.title("Histogram Examble")
plt.xlabel("value")
plt.ylabel("frequency")
plt.show()
```

```python
#pipechart

sizes = [40, 35, 30, 25]
labels = ['python', 'c++', 'html', 'java']

plt.pie(sizes, labels=labels, autopct='%1.1f%%', startangle=140)
plt.title('language popularity share')
plt.show()
```

```python
#data distribution

data = [5, 8, 10, 15, 18, 20, 22, 25, 30]
plt.boxplot(data)
plt.xlabel("box plot example")
plt.ylabel("value")
plt.show()
```

```python
#multiple plots
x = [1, 2, 3, 4, 5]
y1 = [2, 4, 6, 8, 10]
y2 = [1, 3, 5, 7, 9]

plt.subplot(1, 2, 1)
plt.plot(x, y1)
plt.title("plot 1")

plt.subplot(1, 2, 2)
plt.plot(x, y2)
plt.title("plot 2")

plt.show()
```

```python
#legent adds a legent(lble) for multiple plots

x = [1, 2, 3, 4]
plt.plot(x, [1, 4, 9, 16], label='square')
plt.plot(x, [1, 8, 27, 64], label='cube')

plt.legend()
plt.title('legent example')
plt.show()
```

```python
#add gridline for eassier reading of value 

x = [1, 2, 3, 4]
y = [2, 4, 6, 8]

plt.plot(x, y)
plt.grid(True)
plt.title('line plot with grid')
plt.show()
```

```python
#create a new figure (canvas) for multiple plots 

plt.figure(figsize=(6,4))
plt.plot([1, 2, 3], [2, 4, 6])
plt.title("costome figure size")
plt.show()
```

```python
#save the current plot as an image file 

plt.plot([1, 2, 3], [2, 4, 6])
plt.title("save the plot examble")
plt.savefig("plot_example.png")  # Save the plot as a PNG file
plt.show()
```
